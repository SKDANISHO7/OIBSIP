```python
import time
import joblib
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
)

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

start_time = time.time()
```

# Step 1 : Load Dataset

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = [
    iris.target_names[i]
    for i in iris.target
]

print("="*60)
print("IRIS DATASET LOADED SUCCESSFULLY")
print("="*60)
```

# Step 2 : Exploratory Data Analysis

```python
print("Dataset Shape")
print(df.shape)

print("\nFirst Five Rows")
print(df.head())

print("\nDataset Information")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nStatistical Summary")
print(df.describe())

print("\nSpecies Count")
print(df["species"].value_counts())
```

# Step 3 : Data Visualization

```python
sns.set_theme(style="whitegrid")

features = iris.feature_names
```

```python
df[features].hist(
    figsize=(10,8),
    bins=15,
    edgecolor="black"
)

plt.suptitle(
    "Histogram of Iris Features",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
pair = sns.pairplot(
    df,
    hue="species",
    palette="Set2",
    markers=["o","s","D"],
    diag_kind="hist"
)

pair.fig.suptitle(
    "Pairplot of Iris Dataset",
    y=1.02,
    fontsize=16,
    fontweight="bold"
)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[features].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(12,8))

for i, col in enumerate(features):

    plt.subplot(2,2,i+1)

    sns.boxplot(
        data=df,
        x="species",
        y=col,
        hue="species",
        palette="Set2",
        legend=False
    )

    plt.title(col)

plt.tight_layout()
plt.show()
```

# Step 4 : Feature Selection

```python
print("""
Feature Selection Insights

1. Petal Length and Petal Width are the most important features.

2. Sepal Length provides moderate discrimination.

3. Sepal Width is the least informative feature.
""")
```

# Step 5 : Train Test Split

```python
X = df[features]
y = df["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Samples :", X_train.shape[0])
print("Testing Samples :", X_test.shape[0])
```

# Step 6 : Model Training

```python
models = {

    "Logistic Regression":
    LogisticRegression(
        max_iter=300,
        random_state=42
    ),

    "KNN":
    KNeighborsClassifier(
        n_neighbors=5
    ),

    "Decision Tree":
    DecisionTreeClassifier(
        random_state=42
    )
}

results = {}
```

```python
for name, model in models.items():

    print("="*60)
    print(name)
    print("="*60)

    model.fit(X_train,y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    results[name] = accuracy

    print(f"Accuracy : {accuracy*100:.2f}%")

    print(classification_report(
        y_test,
        y_pred
    ))

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=iris.target_names
    )

    plt.figure(figsize=(5,4))

    sns.heatmap(
        cm,
        annot=True,
        cmap="Blues",
        fmt="d",
        xticklabels=iris.target_names,
        yticklabels=iris.target_names
    )

    plt.title(f"Confusion Matrix\n{name}")

    plt.xlabel("Predicted")

    plt.ylabel("Actual")

    plt.show()
```

# Step 7 : Model Comparison

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=list(results.keys()),
    y=list(results.values()),
    hue=list(results.keys()),
    palette="Set2",
    legend=False
)

plt.ylim(0.9,1.0)

plt.ylabel("Accuracy")

plt.title("Accuracy Comparison")

plt.show()
```

# Step 8 : Best Model

```python
best_model_name = max(
    results,
    key=results.get
)

best_model = models[
    best_model_name
]

print("Best Model :", best_model_name)

print(
    "Accuracy :",
    f"{results[best_model_name]*100:.2f}%"
)
```

# Step 9 : Feature Importance

```python
tree = models["Decision Tree"]

importance = pd.DataFrame({

    "Feature":features,

    "Importance":
    tree.feature_importances_

})

print(
    importance.sort_values(
        by="Importance",
        ascending=False
    )
)   
```

# Step 10 : Sample Prediction

```python
sample = X_test.iloc[[0]]

prediction = best_model.predict(sample)

print("Predicted :", prediction[0])

print("Actual :", y_test.iloc[0])
```

# Step 11 : Save Model

```python
joblib.dump(
    best_model,
    "iris_best_model.pkl"
)

print("Model Saved Successfully")
```

# Step 12 : Execution Time

```python
end_time = time.time()

print(
    "Execution Time :",
    round(end_time-start_time,2),
    "seconds"
)
```
